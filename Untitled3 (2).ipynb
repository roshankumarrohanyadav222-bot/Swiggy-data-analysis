```python
# Plot the graph: location vs rating vs delivery time
import seaborn as sns
plt.figure(figsize=(12,8))
sns.scatterplot(data=df,x='rating',y='delivery_time',hue='location',palette='deep',s=60,alpha=0.7)
plt.title('location vs rating vs delivery time',fontsize=16,fontweight='bold')
plt.xlabel("Rating",fontsize=12)
plt.ylabel('delivery time(min)',fontsize=12)
plt.grid(True,linestyle="--",alpha=0.5)
plt.legend(bbox_to_anchor=(1.05,1),loc='upper left',title='Locations')
plt.tight_layout()
plt.show()
```
